# Quadrupolar NMR spectrum simulator

Simulate first/second-order quadrupolar NMR spectra (powder average or
single crystal) from **user-entered parameters** — isotope (or custom I,
γ), magnetic field, the quadrupolar coupling as C_q or ν_Q, and η; plus
optionally lattice vectors, EFG eigenvectors and a field direction. No DFT
calculation is needed; to *compute* EFG tensors with QE-CONVERSE use the
aiidalab-qe plugins shipped in this same package.

Equation numbers in the interface refer to
[T. Arh, *Stability of quantum spin liquids in two dimensions*, doctoral
dissertation, University of Ljubljana (2024)](https://repozitorij.uni-lj.si/IzpisGradiva.php?id=159093&lang=eng).
